```python
import os

import requests
from dotenv import load_dotenv

load_dotenv()

# --- Configuration ---
BASE_URL = "https://kf.kobotoolbox.org"
ASSET_UID = "aLYUSTDALPacjknnFJrmWS"  # FRorm's unique ID from the project's URL
API_TOKEN = os.getenv("API_TOKEN")

API_URL = f"{BASE_URL}/api/v2/assets/{ASSET_UID}/data.json"
HEADERS = {"Authorization": f"Token {API_TOKEN}"}
```

```python
def get_all_submissions():
    all_results = []
    start = 0
    limit = 1000  # Maximum allowed per request

    session = requests.Session()
    session.headers.update({"Authorization": f"Token {API_TOKEN}"})

    while True:
        params = {"limit": limit, "start": start}
        response = session.get(API_URL, params=params)
        response.raise_for_status()
        data = response.json()

        results = data.get("results", [])
        all_results.extend(results)
        print(f"Fetched {len(results)} records (start={start})")

        # Check if there is a 'next' page link
        if not data.get("next"):
            break

        start += limit

    print(f"Total submissions retrieved: {len(all_results)}")
    return all_results
```

```python
submissions_data = get_all_submissions()

# First submission
if submissions_data:
    print(submissions_data[0])
```
